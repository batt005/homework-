16. Дано целое число в диапазоне 20–69, определяющее возраст (в годах).
Вывести строку-описание указанного возраста, обеспечив правильное согласование числа со словом «год», например: 20 — «двадцать лет», 32 —
«тридцать два года», 41 — «сорок один год».

```python
age = int(input())

tens = age // 10
ones = age % 10

tens_words = ["", "", "двадцать", "тридцать", "сорок", "пятьдесят", "шестьдесят"]
ones_words = ["", "один", "два", "три", "четыре", "пять", "шесть", "семь", "восемь", "девять"]

if ones == 0:
    word_age = "лет"
elif ones == 1:
    word_age = "год"
elif 2 <= ones <= 4:
    word_age = "года"
else:
    word_age = "лет"

if ones == 0:
    result = f"{tens_words[tens]} {word_age}"
else:
    result = f"{tens_words[tens]} {ones_words[ones]} {word_age}"

print(result)
```

17. Дано целое число в диапазоне 10–40, определяющее количество учебных заданий по некоторой теме. Вывести строку-описание указанного
количества заданий, обеспечив правильное согласование числа со словами «учебное задание», например: 18 — «восемнадцать учебных заданий»,
23 — «двадцать три учебных задания», 31 — «тридцать одно учебное задание».


```python
num = int(input())

words_10_19 = [
    "десять", "одиннадцать", "двенадцать", "тринадцать", "четырнадцать",
    "пятнадцать", "шестнадцать", "семнадцать", "восемнадцать", "девятнадцать"
]
words_tens = ["", "", "двадцать", "тридцать", "сорок"]
words_ones = ["", "одно", "два", "три", "четыре", "пять", "шесть", "семь", "восемь", "девять"]

if 10 <= num <= 19:
    num_str = words_10_19[num - 10]
    ending = "учебных заданий"
else:
    tens = num // 10
    ones = num % 10

    if ones == 0:
        num_str = words_tens[tens]
        ending = "учебных заданий"
    else:
        num_str = f"{words_tens[tens]} {words_ones[ones]}"
        if ones == 1:
            ending = "учебное задание"
        elif 2 <= ones <= 4:
            ending = "учебных задания"
        else:
            ending = "учебных заданий"

print(f"{num_str} {ending}")
```

18. Дано целое число в диапазоне 100–999. Вывести строку-описание
данного числа, например: 256 — «двести пятьдесят шесть», 814 — «восемьсот четырнадцать».


```python
num = int(input())

hundreds_words = [
    "", "сто", "двести", "триста", "четыреста",
    "пятьсот", "шестьсот", "семьсот", "восемьсот", "девятьсот"
]
tens_words = [
    "", "", "двадцать", "тридцать", "сорок",
    "пятьдесят", "шестьдесят", "семьдесят", "восемьдесят", "девяносто"
]
ones_words = [
    "", "один", "два", "три", "четыре",
    "пять", "шесть", "семь", "восемь", "девять"
]
teens_words = [
    "десять", "одиннадцать", "двенадцать", "тринадцать", "четырнадцать",
    "пятнадцать", "шестнадцать", "семнадцать", "восемнадцать", "девятнадцать"
]

h = num // 100
t = (num // 10) % 10
o = num % 10

result = []

if h > 0:
    result.append(hundreds_words[h])

if t == 1:
    result.append(teens_words[o])
else:
    if t > 1:
        result.append(tens_words[t])
    if o > 0:
        result.append(ones_words[o])

print(" ".join(result))
```

19. В восточном календаре принят 60-летний цикл, состоящий из 12 летних подциклов, обозначаемых названиями цвета: зеленый, красный,
желтый, белый и черный. В каждом подцикле годы носят названия животных: крысы, коровы, тигра, зайца, дракона, змеи, лошади, овцы, обезьяны,
курицы, собаки и свиньи. По номеру года определить его название, если
1984 год — начало цикла: «год зеленой крысы».


```python
year = int(input())

# 1984 год — начало цикла (год зеленой крысы)
# Сдвигаем год относительно начала цикла
offset = (year - 1984) % 60

# Всего 5 цветов, каждый цвет длится 2 года подряд внутри 12-летнего подцикла
colors = ["зеленый", "красный", "желтый", "белый", "черный"]
color_index = (offset % 10) // 2

# Всего 12 животных, меняются каждый год
animals = [
    "крысы", "коровы", "тигра", "зайца", "дракона", "змеи",
    "лошади", "овцы", "обезьяны", "курицы", "собаки", "свиньи"
]
animal_index = offset % 12

# Согласование окончаний цветов в зависимости от животного
# Все животные из списка в родительном падеже, перед ними цвет идет в женском роде
color_word = colors[color_index]
if color_word == "зеленый":
    color_gender = "зеленой"
elif color_word == "красный":
    color_gender = "красной"
elif color_word == "желтый":
    color_gender = "желтой"
elif color_word == "белый":
    color_gender = "белой"
else:
    color_gender = "черной"

print(f"год {color_gender} {animals[animal_index]}")
```
