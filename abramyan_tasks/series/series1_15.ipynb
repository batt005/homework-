4. Дано целое число N и набор из N вещественных чисел. Вывести
сумму и произведение чисел из данного набора.

```python
N = int(input())
total_sum = 0.0
product = 1.0

for _ in range(N):
    num = float(input())
    total_sum += num
    product *= num

print(total_sum)
print(product)
```

5. Дано целое число N и набор из N положительных вещественных
чисел. Вывести в том же порядке целые части всех чисел из данного
набора (как вещественные числа с нулевой дробной частью), а также
сумму всех целых частей.


```python
N = int(input())
total_sum = 0.0

for _ in range(N):
    num = float(input())
    int_part = float(int(num))
    print(int_part)
    total_sum += int_part

print(total_sum)
```

6. Дано целое число N и набор из N положительных вещественных
чисел. Вывести в том же порядке дробные части всех чисел из данного набора (как вещественные числа с нулевой целой частью), а также
произведение всех дробных частей.


```python
N = int(input())
product = 1.0

for _ in range(N):
    num = float(input())
    frac_part = num - int(num)
    print(frac_part)
    product *= frac_part

print(product)
```

7. Дано целое число N и набор из N вещественных чисел. Вывести в
том же порядке округленные значения всех чисел из данного набора (как
целые числа), а также сумму всех округленных значений.


```python
N = int(input())
total_sum = 0

for _ in range(N):
    num = float(input())
    rounded_val = round(num)
    print(rounded_val)
    total_sum += rounded_val

print(total_sum)
```

8. Дано целое число N и набор из N целых чисел. Вывести в том же
порядке все четные числа из данного набора и количество K таких чисел.


```python
N = int(input())
count_even = 0

for _ in range(N):
    num = int(input())
    if num % 2 == 0:
        print(num)
        count_even += 1

print(count_even)
```

9. Дано целое число N и набор из N целых чисел. Вывести в том же
порядке номера всех нечетных чисел из данного набора и количество K
таких чисел.


```python
N = int(input())
count_odd = 0

for i in range(1, N + 1):
    num = int(input())
    if num % 2 != 0:
        print(i)
        count_odd += 1

print(count_odd)
```

10. Дано целое число N и набор из N целых чисел. Если в наборе
имеются положительные числа, то вывести TRUE; в противном случае
вывести FALSE.

```python
N = int(input())
has_positive = False

for _ in range(N):
    num = int(input())
    if num > 0:
        has_positive = True

print(has_positive)
2
```

11. Даны целые числа K, N и набор из N целых чисел. Если в наборе
имеются числа, меньшие K, то вывести TRUE; в противном случае вывести
FALSE.


```python
K = int(input())
N = int(input())
has_less_than_K = False

for _ in range(N):
    num = int(input())
    if num < K:
        has_less_than_K = True

print(has_less_than_K)
1
```

12. Дан набор ненулевых целых чисел; признак его завершения — число 0. Вывести количество чисел в наборе.


```python
count = 0

while True:
    num = int(input())
    if num == 0:
        break
    count += 1

print(count)
```

13. Дан набор ненулевых целых чисел; признак его завершения — число 0. Вывести сумму всех положительных четных чисел из данного набора. Если требуемые числа в наборе отсутствуют, то вывести 0.


```python
total_sum = 0

while True:
    num = int(input())
    if num == 0:
        break
    if num > 0 and num % 2 == 0:
        total_sum += num

print(total_sum)
```

14. Дано целое число K и набор ненулевых целых чисел; признак его
завершения — число 0. Вывести количество чисел в наборе, меньших K.

```python
K = int(input())
count = 0

while True:
    num = int(input())
    if num == 0:
        break
    if num < K:
        count += 1

print(count)
```

15*. Дано целое число K и набор ненулевых целых чисел; признак его
завершения — число 0. Вывести номер первого числа в наборе, большего K. Если таких чисел нет, то вывести 0.

```python
K = int(input())
index = 0
found_index = 0

while True:
    num = int(input())
    if num == 0:
        break
    index += 1
    if num > K and found_index == 0:
        found_index = index

print(found_index)
```
