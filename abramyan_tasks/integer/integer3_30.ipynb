29*. Даны целые положительные числа A, B, C. На прямоугольнике размера A × B размещено максимально возможное количество квадратов со
стороной C (без наложений). Найти количество квадратов, размещенных
на прямоугольнике, а также площадь незанятой части прямоугольника.


```python
A = int(input())
B = int(input())
C = int(input())

count_A = A // C
count_B = B // C
total_squares = count_A * count_B

total_area = A * B
squares_area = total_squares * (C * C)
leftover_area = total_area - squares_area

print(total_squares)
print(leftover_area)
```
