```python
A = float(input("Введите значение A: "))
B = float(input("Введите значение B: "))
C = float(input("Введите значение C: "))

if A < B < C:
    A *= 2
    B *= 2
    C *= 2
else:
    A = -A
    B = -B
    C = -C

print(A)
print(B)
print(C)
```

28. Дан номер года (положительное целое число). Определить количество
дней в этом году, учитывая, что обычный год насчитывает 365 дней, а
високосный — 366 дней. Високосным считается год, делящийся на 4, за
исключением тех годов, которые делятся на 100 и не делятся на 400
(например, годы 300, 1300 и 1900 не являются високосными, а 1200 и 2000
— являются).


```python
year = int(input("Введите номер года: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    days = 366
else:
    days = 365

print(days)
```

29. Дано целое число. Вывести его строку-описание вида «отрицательное
четное число», «нулевое число», «положительное нечетное число» и т. д.

```python
n = int(input("Введите целое число: "))

if n == 0:
    print("нулевое число")
else:
    if n > 0:
        description = "положительное "
    else:
        description = "отрицательное "

    if n % 2 == 0:
        description += "четное число"
    else:
        description += "нечетное число"

    print(description)
```

30. Дано целое число, лежащее в диапазоне 1–999. Вывести его строкуописание вида «четное двузначное число», «нечетное трехзначное число»
и т. д.

```python
n = int(input("Введите число от 1 до 999: "))

if n % 2 == 0:
    description = "четное "
else:
    description = "нечетное "

if 1 <= n <= 9:
    description += "однозначное число"
elif 10 <= n <= 99:
    description += "двузначное число"
else:
    description += "трехзначное число"

print(description)
```
