# Задание 4. Математический спринт
### Суть: Игра на скорость. За отведенное время (например, 60 секунд) нужно решить как можно больше примеров.
Основные требования:
1. Программа генерирует случайные примеры (сложение, вычитание,
умножение).
2. Игрок вводит ответ. Если верно — засчитывается очко и дается следующий пример. Если неверно — пример повторяется или пропускается (на твой выбор).
3. В консоли должно быть видно оставшееся время (или таймер просто работает в фоне).
4. По истечении времени игра останавливается и выводит итоговый счет и статистику (сколько верных/неверных).

```python
import random
import time

time_limit = 60
correct_answers = 0
wrong_answers = 0

print(f"Математический спринт! У вас есть {time_limit} секунд, чтобы решить как можно больше примеров.")
input("Нажмите Enter, чтобы начать...")

start_time = time.time()

while True:
    elapsed_time = time.time() - start_time
    remaining_time = int(time_limit - elapsed_time)

    if remaining_time <= 0:
        break

    num1 = random.randint(1, 10)
    num2 = random.randint(1, 10)
    operation = random.choice(["+", "-", "*"])

    if operation == "+":
        correct_result = num1 + num2
    elif operation == "-":
        correct_result = num1 - num2
    else:
        correct_result = num1 * num2

    print(f"\n[Осталось времени: {remaining_time} сек.]")

    try:
        user_answer = int(input(f"Сколько будет {num1} {operation} {num2}? "))
    except ValueError:
        wrong_answers += 1
        print("Ошибка! Нужно вводить только числа. Пример пропущен.")
        continue

    if time.time() - start_time >= time_limit:
        break

    if user_answer == correct_result:
        correct_answers += 1
        print("Верно!")
    else:
        wrong_answers += 1
        print(f"Неверно! Правильный ответ: {correct_result}")

print("\n" + "="*30)
print("ВРЕМЯ ВЫШЛО!")
print(f"Итоговый счет (верно): {correct_answers}")
print(f"Статистика ответов: {correct_answers} верных, {wrong_answers} неверных.")
print("="*30)
```
