# Задание 2. Виселица (Угадай слово)
### Суть: Игрок угадывает скрытое слово, называя буквы. За каждую ошибку отнимается «жизнь».
Основные требования:
1. Заранее подготовленный массив/список слов. Случайный выбор одного.
2. Слово отображается в виде прочерков или нижних подчеркиваний (например: _ _ _ _ _).
3. При вводе буквы программа проверяет, есть ли она в слове. Если да — открывает все такие буквы. Если нет — вычитает жизнь.
4. Вывод сообщения о победе или поражении.


```python
import random

words = ["программирование", "компьютер", "питон", "алгоритм", "разработка", "скрипт"]
secret_word = random.choice(words)

guessed_letters = []
lives = 6

while lives > 0:
    display_word = ""
    for letter in secret_word:
        if letter in guessed_letters:
            display_word += letter + " "
        else:
            display_word += "_ "

    print(display_word.strip())
    print(f"Осталось жизней: {lives}")

    if "_" not in display_word:
        print("Поздравляем!Вы победили!")
        break

    guess = input("Введите букву: ")

    if len(guess) != 1 or not guess.isalpha():
        print("Пожалуйста, введите ровно одну букву.")
        continue

    if guess in guessed_letters:
        print("Вы уже вводили эту букву.")
        continue

    guessed_letters.append(guess)

    if guess in secret_word:
        print("Правильно!")
    else:
        lives -= 1
        print("Неверно! Вы теряете жизнь.")
else:
    print(f"Вы проиграли! Было загадано слово: {secret_word}")
```
