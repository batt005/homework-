# Задание 1. Камень, Ножницы, Бумага (на победу)
### Суть: Классическая игра против компьютера, но с условием: играть нужно до 3-х побед.
Основные требования:
1. Компьютер случайным образом выбирает один из трех вариантов.
2. Игрок вводит свой вариант
3. Вывод результата раунда (Победа / Поражение / Ничья) и общего счета.
4. Игра идет в цикле, пока кто-то не наберет 3 очка.
5. В конце предложение сыграть еще раз.


```python
import random
player_score = 0
pc_score = 0
while True:
    computer = random.randint(1, 3)
    player = int(input("Введите свой вариант (1 - камень, 2 - ножницы, 3 - бумага): "))
    if computer == player:
        print("Ничья")
    elif (player == 1 and computer == 2) or (player == 2 and computer == 3) or (player == 3 and computer == 1):
        player_score += 1
        print("Победа")
    else:
        pc_score += 1
        print("Поражение")
    print(f"Текущий счет: Игрок {player_score} - {pc_score} Компьютер")
    if player_score == 3 or pc_score == 3:
        if player_score == 3:
            print("Вы победили!")
        else:
            print("Вы проиграли машине. Ехеехеххехех")
        count = input("Хотите сыграть еще раз? (yes/no): ")
        if count == "yes":
            player_score = 0
            pc_score = 0
            print("Новая игра")
        else:
            print("Пока-пока!")
            break
```
