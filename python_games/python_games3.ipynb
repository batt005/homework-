# Задание 3. Угадай число
### Суть: Программа загадывает число от 1 до 100. Пользователь пытается угадать это число.
Основные требования:
1. После попытки угадать программа сообщает один из вариантов:
  - Не угадал! Загаданное число БОЛЬШЕ
  - Не угадал! Загаданное число МЕНЬШЕ
  - Поздравляем ты угадал число!
2. Программа считает количество неудачных попыток и выводит счет в конце игры.
3. \* Улучшите игру так чтобы пользователь мог сам указывать диапазон загаднного числа. Диапазон не должен быть меньше 10. И не должен быть отрицательным.
4. \** Определить минимальное необходимое число попыток для нахождения загаданного числа. Если пользователь не укладывается в это количество попыток - то он проигрывает. (Задача учитывается только при выполнении пункта 3.

```python
import random
import math
while True:
    min_num = int(input("Введите начало диапазона: "))
    max_num = int(input("Введите конец диапазона: "))
    if min_num < 0 or max_num < 0:
        print("Диапазон не должен быть отрицательным. Попробуйте еще раз")
        continue
    diff = max_num - min_num
    if diff < 10:
        print("Разница между числами должна быть не меньше 10. Попробуйте еще раз")
        continue
    break
secret_number = random.randint(min_num, max_num)
max_attempts = math.ceil(math.log2(diff + 1))
wrong_attempts = 0
print(f"\nЗагадано число от {min_num} до {max_num}.")
print(f"У вас есть {max_attempts} попыток.\n")
while wrong_attempts < max_attempts:
    user_guess = int(input("Введите ваше число: "))
    if user_guess == secret_number:
        print("Поздравляем ты угадал число!")
        print(f"Количество неудачных попыток: {wrong_attempts}")
        break
    elif user_guess < secret_number:
        print("Не угадал! Загаданное число БОЛЬШЕ")
        wrong_attempts += 1
    else:
        print("Не угадал! Загаданное число МЕНЬШЕ")
        wrong_attempts += 1
else:
    print("\nВы проиграли! Закончились попытки.")
    print(f"Было загадано число: {secret_number}")
    print(f"Количество неудачных попыток: {wrong_attempts}")
```
